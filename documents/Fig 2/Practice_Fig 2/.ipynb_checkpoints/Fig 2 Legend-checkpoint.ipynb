```python
#Plus

import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.legend_handler import HandlerTuple
import seaborn as sns

# Function to create and plot only the customized legend
def plot_custom_legend(vehicle_groups, legend_ncol=3, legend_loc='upper center', legend_bbox=(0.5, 0.5), legend_fontsize=12):
    # Set the plot style
    sns.set(style="whitegrid")

    # Set the font to Times New Roman
    plt.rcParams["font.family"] = "Times New Roman"

    # Create a dummy figure
    fig, ax = plt.subplots(figsize=(8, 2))  # Adjust the figure size as needed
    
    # Define colors using Seaborn's color palette
    colors = sns.color_palette("husl", len(vehicle_groups))
    
    # Create dummy scatter points and lines for the legend
    handles = []
    labels = []
    for idx, (vehicle_type, fuel_type) in enumerate(vehicle_groups):
        scatter = Line2D([], [], color=colors[idx], marker='+', linestyle='None', markersize=10)
        line = Line2D([], [], color=colors[idx], linewidth=1.5)
        handles.append((scatter, line))  # Tuple to combine scatter and line
        labels.append(f'{vehicle_type} - {fuel_type}')

    # Customize the legend
    legend = ax.legend(handles, labels, loc=legend_loc, bbox_to_anchor=legend_bbox,
                       frameon=False, ncol=legend_ncol, fontsize=legend_fontsize,
                       handler_map={tuple: HandlerTuple(ndivide=None)})

    # Hide the axes
    ax.axis('off')
    
    # Adjust layout and display
    plt.tight_layout()
    plt.show()

# Manually input vehicle type and fuel type groups
vehicle_fuel_groups = [
    ('Passenger Car', 'Gasoline'),
    ('Passenger Car', 'Ethanol (E-85)'),
    ('Light Commercial Truck', 'Gasoline'),
    ('Passenger Truck', 'Gasoline'),
    ('Motorcycle', 'Gasoline'),
    ('Passenger Car', 'Diesel Fuel'),
    ('Single Unit Short-haul Truck', 'Diesel Fuel'),
    ('Passenger Truck', 'Ethanol (E-85)'),
    ('Combination Long-haul Truck', 'Diesel Fuel'),
    ('Single Unit Short-haul Truck', 'Gasoline'),
    ('Light Commercial Truck', 'Ethanol (E-85)'),
    ('Passenger Car', 'Electricity')
]

# Call the function to plot only the customized legend
plot_custom_legend(vehicle_fuel_groups, legend_ncol=3, legend_loc='upper center', legend_bbox=(0.5, 0.5), legend_fontsize=12)
```
